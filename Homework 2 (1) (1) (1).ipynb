## Q2. Combine order details and summarize category sales

Merge `orders` with `products` using `product_id`, keeping every order. Then:

## Q3. Build a salesman-by-customer purchase table

Write a function `sales_matrix(orders)` that returns a DataFrame with:
- One row per `salesman_id` and one column per `customer_id`.
- The **sum** of `purch_amt` for each salesman-customer pair.
- `0` for pairs with no orders.

Sort the row and column labels in ascending order. Keep the input unchanged. Call the function with the supplied table and print the result. Repeated salesman-customer pairs must be combined.

```python
orders = pd.DataFrame({
    "ord_no": [70001, 70009, 70002, 70004, 70007, 70005, 70008, 70010],
    "purch_amt": [150.50, 270.65, 65.26, 110.50, 948.50, 2400.60, 5760.00, 1983.43],
    "customer_id": [3005, 3001, 3002, 3009, 3005, 3007, 3002, 3004],
    "salesman_id": [5002, 5005, 5001, 5003, 5002, 5001, 5001, 5006],
})

# Your code here
def sales_matrix(orders):
    matrix = orders.pivot_table(
        index="salesman_id",
        columns="customer_id",
        values="purch_amt",
        aggfunc="sum",
        fill_value=0)
    matrix = matrix.sort_index(axis=0, ascending=True)
    matrix = matrix.sort_index(axis=1, ascending=True)
    return matrix
result = sales_matrix(orders)
print(result)
```
